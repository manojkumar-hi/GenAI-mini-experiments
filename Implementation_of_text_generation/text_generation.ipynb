# Objective
Implement text generation using the pre-trained HuggingFace Transformer model DistilGPT-2.

The aim is to generate a news-style paragraph from a user-provided headline or topic.

# Method
Load the DistilGPT-2 model from HuggingFace, accept a headline using `input()`, generate a paragraph, and print the result.

```python
# Install required packages in the notebook environment if needed
# !pip install flask flask-cors transformers torch

from transformers import pipeline

text_generator = pipeline('text-generation', model='distilgpt2')
```

# Code
Take user input and generate a news article paragraph.

```python
headline = input('Enter a news headline or topic: ').strip()

if not headline:
    print('Headline cannot be empty.')
else:
    prompt = f'Write a professional news article paragraph based on this headline: {headline}\nNews article:'
    result = text_generator(
        prompt,
        max_length=150,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7,
        repetition_penalty=1.2,
        no_repeat_ngram_size=3,
    )
    generated_article = result[0]['generated_text']
    print('\nGenerated Article:\n')
    print(generated_article)
```

# Result
The notebook prints a generated article draft based on the entered headline.

# Conclusion
This experiment demonstrates how a pre-trained Transformer can be used for simple news text generation.
